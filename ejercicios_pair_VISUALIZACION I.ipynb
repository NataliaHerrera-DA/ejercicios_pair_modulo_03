```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None) 
```

```python
import warnings
warnings.filterwarnings('ignore', category=FutureWarning)
```

```python
df = pd.read_csv('world_data_full_apply_nulos.csv', index_col=0)
df.head(5)
```

1. ¿Cómo se distribuye la tasa de natalidad en el conjunto de datos? ¿Hay algún patrón evidente?


```python
sns.histplot(data=df, x='birth_rate', kde=True)

plt.title('Gráfica tasa de natalidad')
plt.xlabel('Tasa_Natalidad')
```

```python
sns.boxplot(data=df, x='birth_rate')

plt.xlabel('Tasa_Natalidad')
```

2. ¿Existe una relación aparente entre la tasa de mortalidad infantil y la esperanza de vida? Proporciona una breve descripción de los patrones que observes.


```python
df.columns
```

```python
sns.scatterplot(data=df, x='infant_mortality', y='life_expectancy') 

plt.title('Relación entre la tasa de mortalidad infantil y esperanza de vida')
plt.xlabel('Tasa_Mortalidad_Infantil')
plt.ylabel('Esperanza_Vida')
```

Se puede observar que la esperanza de vida es mayor mientras la mortalidad infantil es más baja. A medida que la mortalidad infantil disminuye, la esperanza de vida aumenta. 

```python
sns.scatterplot(data=df, x='infant_mortality', y='life_expectancy', hue='birth_rate') 

plt.title('Relación entre la tasa de mortalidad infantil y esperanza de vida')
plt.xlabel('Tasa_Mortalidad_Infantil')
plt.ylabel('Esperanza_Vida')
```

3. ¿Cuál es la mediana de la esperanza de vida en Europa? Proporciona un rango intercuartil aproximado y verifica si existen valores atípicos.

```python
sns.boxplot(data=df, x='life_expectancy', y='continent')

plt.title('Mediana de la tasa de vida por continente')
plt.xlabel('Esperanza_Vida')
plt.ylabel('Continente')
```

```python
sns.boxplot(data=df[df['continent'] == 'Europe'], x='life_expectancy')

plt.title('Mediana de la tasa de vida en Europa')
plt.xlabel('Esperanza_Vida')
```

Según la gráfica, la mediana es 81. No hay nigun dato atípico y el rango intercuartil es entre 76 y 82.

```python
lista_columnas = ['infant_mortality', 'life_expectancy', 'density', 'urban_population']

fig, axes = plt.subplots(nrows = 2, ncols = 2, figsize = (15, 12))

axes = axes.flatten()

plt.suptitle('Relaciones visibles entre la tasa de natalidad y las siguientes variables: tasa de mortalidad infantil, esperanza de vida, densidad de población y población urbana')

axes[0].set_title("Relación entre la tasa de natalidad y la tasa de mortalidad infantil")
axes[1].set_title("Relación entre la tasa de natalidad y la esperanza de vida")
axes[2].set_title("Relación entre la tasa de natalidad y la densidad de población")
axes[3].set_title("Relación entre la tasa de natalidad y la población urbana")

axes[0].set_xlabel("Tasa_Mortalidad")
axes[0].set_ylabel("Tasa_Natalidad")

axes[1].set_xlabel("Esperanza_Vida")
axes[1].set_ylabel("Tasa_Natalidad")

axes[2].set_xlabel("Densidad_Población")
axes[2].set_ylabel("Tasa_Natalidad")

axes[3].set_xlabel("Población_Urbana")
axes[3].set_ylabel("Tasa_Natalidad")


for a, v in enumerate(lista_columnas):
    sns.scatterplot(data=df, x=v, y='birth_rate', ax=axes[a])





```
