```python
df_u.rename(columns=df_dic, inplace=True)
```

```python
df_u
```

```python
#solución con función: 

columnas = df_u.columns.to_series()

columnas = (
    columnas
    .str.lower()                                 # Todo a minúsculas
    .str.replace(r'[\n:/]', ' ', regex=True)     # Reemplazar '\n', ':', '/' con espacio
    .str.replace(r'[\(\)]', '', regex=True)      # Eliminar paréntesis '(', ')'
    .str.replace(r' ', '_', regex=True)          # Reemplazar espacios con guion bajo
    .str.replace(r'_{2,}', '_', regex=True)      # Eliminar guiones bajos múltiples
    .str.strip('_')                              # Eliminar guiones bajos al inicio/final
)

df_u.columns = columnas
```
