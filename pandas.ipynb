### PANDAS


¿Qué diferencia hay entre una Serie y un DataFrame?

1. La Serie 

        - Una Serie es el objeto fundamental unidimensional de Pandas.

        - Es esencialmente una columna de datos.
      
        - Contiene una matriz de datos (los valores) y un índice asociado, que son etiquetas que dan nombre a cada fila (por defecto, son números enteros desde 0).

        - Se asemeja a una lista de Python que ha sido mejorada con etiquetas de acceso rápido (el índice).

2. El DataFrame

        - El DataFrame es el objeto principal bidimensional de Pandas.

        - Es la estructura de datos más utilizada y representa una tabla con filas y columnas.

        - Puede verse como una colección de objetos Series que comparten el mismo índice de fila.

        - Es la forma en que Pandas representa la mayoría de los datos del mundo real, como hojas de cálculo o tablas de bases de datos.

- Tiene dos índices:

        - Índice de fila (eje 0): Las etiquetas de las filas.

        - Nombres de columna (eje 1): Las etiquetas de las columnas.

#### En resumen: el DataFrame es la tabla, y cada columna de esa tabla es una Serie.

```python
ind = list(indices)
```

```python
val = list(valores)
```

```python
dic = {} 

for v, i in zip(ind, val):
    dic[v] = i
    print(f"par agregado {i} y {v}")
```
